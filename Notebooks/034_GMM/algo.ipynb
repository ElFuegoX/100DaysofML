```python
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt

X, y = make_blobs(
    n_samples=400, 
    centers=4, 
    n_features=2, 
    cluster_std=1.2, 
    random_state=42
)

plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis', s=40, alpha=0.8)
plt.title("Dataset généré avec make_blobs (4 clusters)")
plt.show()
```

```python
# Gaussian Mixture Model 
from sklearn.mixture import GaussianMixture
gmm = GaussianMixture(n_components=4, random_state=42)
gmm.fit(X)
y_gmm = gmm.predict(X)             
proba = gmm.predict_proba(X)        
```

```python
from sklearn.metrics import silhouette_score, adjusted_rand_score
for i in range(2, 10):
    gmm = GaussianMixture(n_components=i, random_state=42)
    gmm.fit(X)
    y_gmm = gmm.predict(X)
    print(f"Nombre de clusters: {i}")
    print("Silhouette Score (GMM) :", silhouette_score(X, y_gmm))
    print("Adjusted Rand Index (GMM) :", adjusted_rand_score(y, y_gmm))
    print("BIC du GMM :", gmm.bic(X))
    print("AIC du GMM :", gmm.aic(X))
    print("-" * 50)
```
